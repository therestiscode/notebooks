# Two colons. Twenty gigabytes.

Five thousand numbers take 40 kilobytes. One subtraction of them from themselves takes 200 megabytes.
The first three cells are the ones in the video, numbered as they were. The rest is the evidence.

## The video

```python
import numpy as np
a = np.arange(5_000, dtype=float)
print(a.nbytes)
```

```python
print((a[:, None] - a[None, :]).nbytes)
```

```python
a = np.arange(50_000, dtype=float)
print(f"{len(a) * len(a) * 8:,}")
```

The 50,000 in the last cell is arithmetic: the length, times itself, times eight. It was never allocated, here or in the video. Everything below runs at 5,000.

## The colons are free

Both are views of `a`. Neither copies a number.

```python
a = np.arange(5_000, dtype=float)
col, row = a[:, None], a[None, :]
print(col.base is a, row.base is a)
```

## The minus is the bill

One answer for every pair: a row of 5,000 answers for each of the 5,000 numbers, eight bytes each. Every floating-point error is turned into an exception first, and none is raised.

```python
np.seterr(all="raise")
pairs = col - row
print(pairs.shape)
print(f"{pairs.size:,}", pairs.itemsize)
print(pairs.nbytes, pairs.nbytes // a.nbytes)
```

It is a new array, not a view, and the answers are right.

```python
print(pairs.base is None)
print(pairs[1, 0], pairs[0, 1])
```
